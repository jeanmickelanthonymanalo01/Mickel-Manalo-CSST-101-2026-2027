3. Numerical Errors

```r
x <- 10
y <- 3


add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y

print(add) #+
print(sub) #-
print(mul) #*
print(div) #/
print(mod) #%%
print(exp) #^
```
